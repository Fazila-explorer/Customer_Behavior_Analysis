```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform (lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns= df.columns.str.lower()
df.columns= df.columns.str.replace(' ','_')
df= df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate customer IDs:", df["customer_id"].duplicated().sum())
```

```python
# create a column age_group
labels = ['Young Adult','Adult','Middle-aged','Senior']
df['age_group']= pd.qcut(df['age'], q=4, labels= labels)
```

```python
df.groupby("age_group", observed=True)["age"].agg(["min", "max"])
```

```python
df[['age','age_group']].head(10)
```

```python
# create column purchase_frequecncy_days
frequency_mapping = { 
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
'Quarterly': 90,
'Bi-Weekly': 14,
'Annually': 365,
'Every 3 months': 90
}
df['purchase_frequency_days']= df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
# Spend tier: 
df["spend_tier"] = pd.cut(df["purchase_amount"], bins=[0, 40, 70, 100], labels=["Low", "Mid", "High"])

df["is_subscribed"] = (df["subscription_status"] == "Yes").astype(int)
df["got_discount"] = (df["discount_applied"] == "Yes").astype(int)

df[["purchase_amount", "spend_tier", "subscription_status", "is_subscribed"]].head()
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
(df['discount_applied']== df['promo_code_used']).all()
```

```python
df= df.drop('promo_code_used',axis=1)
```

```python
df.columns
```

```python
total_revenue = df["purchase_amount"].sum()

print("Customers:            ", len(df))
print("Total revenue:        $", total_revenue)
print("Average purchase:     $", round(df["purchase_amount"].mean(), 2))
print("Average rating:       ", round(df["review_rating"].mean(), 2))
print("Subscribers:          ", round(df["is_subscribed"].mean() * 100, 1), "%")
print("Discounted purchases: ", round(df["got_discount"].mean() * 100, 1), "%")
```
